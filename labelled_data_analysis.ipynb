```python
import pickle

lfw_race_labels1 = []
lfw_race_labels2 = []
lfw_race_labels3 = []

with open('data/lfw-labels-dlib-mtcnn.pickle', 'rb') as pfile:
    lfw_race_labels2 = pickle.load(pfile)

with open('data/lfw-labels-opencv-ssd.pickle', 'rb') as pfile:
    lfw_race_labels1 = pickle.load(pfile)

with open('data/lfw-labels-retinaface.pickle', 'rb') as pfile:
    lfw_race_labels3 = pickle.load(pfile)
```

```python
len(lfw_race_labels1.keys()) # detector backends - opencv, ssd -- 30 images did not pass detection stage.
```

```python
len(lfw_race_labels2.keys()) # detector backends - dlib, mtcnn -- 1 image did not pass detection stage.
```

```python
len(lfw_race_labels3.keys()) # detector backends - retinaface -- all images passed detection stage.
```

```python
import pandas as pd

df1 = pd.DataFrame([{
        "image": image.split('/')[1].split('-')[0], 
        "gender": dict(pd.Series([lfw_race_labels1[key][0]['dominant_gender'] for key in lfw_race_labels1 if image.split('/')[1].split('-')[0] in key and 'opencv' in key]).value_counts()),
        "race": dict(pd.Series([lfw_race_labels1[key][0]['dominant_race'] for key in lfw_race_labels1 if image.split('/')[1].split('-')[0] in key and 'opencv' in key]).value_counts())
    } for image in lfw_race_labels1 if image.split('/')[1].split('-')[0] in image and 'opencv' in image]
).drop_duplicates(subset=['image'])

df2 = pd.DataFrame([{
    "image": image.split('/')[1].split('-')[0], 
    "gender": dict(pd.Series([lfw_race_labels2[key][0]['dominant_gender'] for key in lfw_race_labels2 if image.split('/')[1].split('-')[0] in key and 'dlib' in key]).value_counts()),
    "race": dict(pd.Series([lfw_race_labels2[key][0]['dominant_race'] for key in lfw_race_labels2 if image.split('/')[1].split('-')[0] in key and 'dlib' in key]).value_counts())
    } for image in lfw_race_labels2 if image.split('/')[1].split('-')[0] in image and 'dlib' in image]
).drop_duplicates(subset=['image'])

df3 = pd.DataFrame([{
    "image": image.split('/')[1].split('-')[0], 
    "gender": dict(pd.Series([lfw_race_labels3[key][0]['dominant_gender'] for key in lfw_race_labels3 if image.split('/')[1].split('-')[0] in key and 'retinaface' in key]).value_counts()),
    "race": dict(pd.Series([lfw_race_labels3[key][0]['dominant_race'] for key in lfw_race_labels3 if image.split('/')[1].split('-')[0] in key and 'retinaface' in key]).value_counts())
    } for image in lfw_race_labels3 if image.split('/')[1].split('-')[0] in image and 'retinaface' in image]
).drop_duplicates(subset=['image'])
```
